## 4. Data Cleaning – Handling Missing Values

Missing values can affect the accuracy of data analysis. We identify missing values using `df.isna().sum()`.

The following cleaning operations are performed:

- Age: Missing values are filled using the median age of passengers with the same passenger class and gender.
- Embarked: Missing values are filled using the most frequent boarding location.
- Deck: The column is removed because it contains a large number of missing values.
- Embark Town: Missing values are filled using the most frequent town.

These methods help prepare the dataset for further analysis.

```python
df.isna().sum()
```

```python
df["age"] = df["age"].fillna(df.groupby(["pclass", "sex"])["age"].transform("median"))
```

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

```python
df = df.drop(columns=["deck"])
```

```python
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
```

```python
df.isna().sum()
```

## 6. Exploratory Data Analysis – Survival Analysis

In this section, we analyze passenger survival rates based on three important factors:

1. Gender
2. Passenger Class
3. Age Group

The `groupby()` function is used to group passengers, and the `mean()` function calculates their survival rates.

### 6.1 Survival Rate by Gender

This analysis compares the survival rates of male and female passengers.

We group the passengers by gender and calculate the average survival value.

**Finding:** Female passengers had a survival rate of approximately 74%, compared to 19% for male passengers.

```python
df.groupby("sex")["survived"].mean().round(2)
```

### 6.2 Survival Rate by Passenger Class

This analysis compares survival rates among first, second, and third-class passengers.

We group passengers by `pclass` and calculate the average survival rate and the number of passengers in each class.

**Finding:** First-class passengers had the highest survival rate at 63%, followed by second class at 47% and third class at 24%.


```python
df.groupby("pclass")["survived"].agg(["mean", "count"]).round(2)
```

### 6.3 Survival Rate by Age Group

In this analysis, passengers are divided into five age groups:

- Child
- Teen
- Adult
- Middle-aged
- Senior

The `pd.cut()` function is used to create age groups, and the survival rate is calculated for each group.

**Finding:** Children had the highest survival rate among the age groups at 58%, while senior passengers had the lowest survival rate at 23%.

```python
df["age_band"] = pd.cut(df["age"], [0, 12, 18, 40, 60, 100],
                        labels=["Child", "Teen", "Adult", "Middle-aged", "Senior"])

df.groupby("age_band", observed=True)["survived"].mean().round(2)
```
